# Six Chunking Methods for RAG

A beginner-friendly notebook with short Python examples. Each code cell demonstrates one way to divide text into chunks. You can run the cells from top to bottom.

**RAG** searches relevant information and gives it to an AI to help answer a question. Chunking prepares documents by splitting them into searchable pieces.


## Start with a small sample

This sample has three short sections. The helper prints the chunks so you can see what each method does.


```python
text = """# Monitoring
Collect service metrics and logs. Set useful alerts.

# Incidents
Assess customer impact. Assign an incident lead.

# Costs
Review cloud usage. Set a budget alert."""

def show(chunks):
    for number, chunk in enumerate(chunks, 1):
        print(f"{number}. {chunk!r}")
```

## 1. Fixed-size chunking

Cuts text into pieces of the same character length. It is simple, but a cut can happen in the middle of a sentence. Change `size` to make the pieces longer or shorter.


```python
size = 55
chunks = [text[i:i + size] for i in range(0, len(text), size)]
show(chunks)
```

## 2. Content-defined (structure-aware) chunking

Uses the document’s headings to create one chunk per section. This works well when documents have clear headings. If one section is very long, it may need to be split again.


```python
sections = [part.strip() for part in text.split("# ") if part.strip()]
chunks = ["# " + section for section in sections]
show(chunks)
```

## 3. Delimiter-based chunking

Splits text at a chosen marker. Here, the marker is a full stop followed by a space, so each sentence becomes a piece. Delimiters are useful for consistently formatted text such as logs or records.


```python
chunks = [sentence.strip() for sentence in text.replace("\n", " ").split(". ") if sentence.strip()]
show(chunks)
```

## 4. Recursive chunking

Tries natural boundaries in order: paragraphs, then sentences, then words. It only uses smaller boundaries when a piece is too long. This short example shows the idea and uses a character limit.


```python
def split_recursively(piece, limit):
    if len(piece) <= limit: return [piece]
    for separator in ["\n\n", ". ", " "]:
        if separator in piece:
            return sum((split_recursively(p, limit) for p in piece.split(separator) if p), [])
    return [piece[i:i + limit] for i in range(0, len(piece), limit)]

show(split_recursively(text, 55))
```

## 5. Overlapping chunking

Makes neighboring chunks share a little text. The repeated words help preserve context when an important idea falls near a chunk boundary. Too much overlap repeats unnecessary text.


```python
size, overlap = 55, 10
step = size - overlap
chunks = [text[i:i + size] for i in range(0, len(text), step)]
show(chunks)
```

## 6. Paragraph-based chunking

Keeps each paragraph together by splitting at blank lines. This is a natural fit for articles and reports where paragraphs express complete thoughts.


```python
chunks = [paragraph.strip() for paragraph in text.split("\n\n") if paragraph.strip()]
show(chunks)
```

## Which method should I try first?

- **Headings in the document?** Try content-defined chunking.
- **Paragraphs are meaningful?** Try paragraph-based chunking.
- **Repeated separators in logs or records?** Try delimiter-based chunking.
- **Mixed formatting?** Try recursive chunking.
- **Need a quick baseline?** Try fixed-size chunking.
- **Context gets lost at boundaries?** Add a little overlap.

These examples use character counts for simplicity. In a real RAG system, teams often tune chunk sizes by checking whether the system retrieves the right information for representative questions.
