```python
%pip install -q langchain-openai python-dotenv "pydantic>=2,<3"

import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()


MODEL_NAME = os.getenv("OPENAI_MODEL", "gpt-4.1-mini")
```

### 2. Create a reusable prompt template

This cell creates a reusable prompt template with variables so the same workflow can handle different inputs consistently.

**Expected result:** The rendered prompt is sent to the model and the resulting text is displayed. Read the output before continuing to the next cell.

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {department} assistant. Be concise and do not invent policies."),
    ("human", "Question: {question}")
])
chain = prompt | ChatOpenAI(model=MODEL_NAME, temperature=0) | StrOutputParser()
print(chain.invoke({"department":"HR", "question":"Draft a polite request for two days of leave."}))
```

## Expected result

The model should return an output that follows the requested scope and format. During training, compare the actual response with the requirement instead of assuming that a fluent response is correct.

## Parameter experiment
